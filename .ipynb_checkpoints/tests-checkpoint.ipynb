```python
from qiskit import QuantumCircuit, QuantumRegister,ClassicalRegister,transpile
from qiskit.circuit.library import CDKMRippleCarryAdder
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
from qiskit.circuit.library import MCXGate
```

```python
N = 4  # number of bits

# registers
a = QuantumRegister(N, "a")
b = QuantumRegister(N, "b")
f = QuantumRegister(1, "f")
m = ClassicalRegister(2,"m")
carry = QuantumRegister(2, "carry")

qc = QuantumCircuit(a, b, f,carry,m)
controls = [b[j] for j in range(N)]
mcx = MCXGate(len(controls))  # reusable gate
# Example: a = 5 (00000101), b = 3 (00000011)
qc.x(a[0])
qc.x(a[2])

#encode 3 in b
qc.x(b[0])
qc.x(b[1])
#2s complement
qc.x(b)
qc.x(carry[0])
# CDKM adder
adder = CDKMRippleCarryAdder(N, kind="full")


qc.append(adder, [carry[0]]+ a[:] + b[:] + [carry[1]])
# qc.append(mcx, controls + [f[0]])

qc.append(adder.inverse(), [carry[0]]+ a[:] + b[:] + [carry[1]])
qc.measure(carry,m)

qc.draw()
```

```python
# Transpile for simulator
simulator = AerSimulator()
qc = transpile(qc, simulator)
results = simulator.run(qc).result()
counts = results.get_counts()
plot_histogram(counts, title="Histogram")

```
